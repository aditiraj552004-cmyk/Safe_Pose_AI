```python
import os
import numpy as np
import pandas as pd

# ===============================
# PATHS
# ===============================

CSV_PATH = "csv_data/landmarks.csv"

OUTPUT_FOLDER = "processed_data"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

SEQUENCE_LENGTH = 30

# ===============================
# LOAD CSV
# ===============================

print("Loading CSV...")

df = pd.read_csv(CSV_PATH)

print(df.head())

# ===============================
# GROUP BY SEQUENCE
# ===============================

X = []
y = []

groups = df.groupby("sequence")

print("Total sequences :", len(groups))

# ===============================
# CREATE SEQUENCES
# ===============================

for sequence_name, group in groups:

    group = group.sort_values("frame")

    label = group["label"].iloc[0]

    landmarks = group.drop(
        columns=["sequence", "frame", "label"]
    ).values

    total_frames = len(landmarks)

    # Sliding window
    for start in range(0, total_frames - SEQUENCE_LENGTH + 1):

        end = start + SEQUENCE_LENGTH

        sequence = landmarks[start:end]

        X.append(sequence)

        y.append(label)

X = np.array(X)

y = np.array(y)

print("X Shape :", X.shape)
print("y Shape :", y.shape)

# ===============================
# SAVE
# ===============================

np.save(
    os.path.join(OUTPUT_FOLDER, "X.npy"),
    X
)

np.save(
    os.path.join(OUTPUT_FOLDER, "y.npy"),
    y
)

print()

print("==============================")
print("Preprocessing Completed")
print("==============================")
print("Saved : processed_data/X.npy")
print("Saved : processed_data/y.npy")
print("Samples :", len(X))
```
